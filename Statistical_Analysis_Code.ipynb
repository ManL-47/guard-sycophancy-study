# T-Test

```python
import numpy as np
from scipy import stats

# Enter your five scenario-level sycophancy values
sycophancy_values = np.array([
   1,
   1,
 -0.545454546,
 0.666666666,
 0.818181818
])

n = len(sycophancy_values)
mean_sv = np.mean(sycophancy_values)
sd_sv = np.std(sycophancy_values, ddof=1)
se_sv = sd_sv / np.sqrt(n)

# One-sample t-test against 0
# H0: mean sycophancy value <= 0
# H1: mean sycophancy value > 0
t_stat = mean_sv / se_sv
df = n - 1
p_one_tailed = stats.t.sf(t_stat, df)

# 95% CI
t_crit = stats.t.ppf(0.975, df)
ci_low = mean_sv - t_crit * se_sv
ci_high = mean_sv + t_crit * se_sv

# Cohen's d
cohens_d = mean_sv / sd_sv

print(f"N = {n}")
print(f"Mean sycophancy value = {mean_sv:.3f}")
print(f"SD = {sd_sv:.3f}")
print(f"95% CI = [{ci_low:.3f}, {ci_high:.3f}]")
print(f"t({df}) = {t_stat:.3f}")
print(f"One-tailed p-value = {p_one_tailed:.6f}")
print(f"Cohen's d = {cohens_d:.3f}")

if p_one_tailed < 0.05 and mean_sv > 0:
    print("Result: Significant movement toward the inducement target.")
else:
    print("Result: No significant movement toward the inducement target.")
```

B-H FDR Correction

```python
from statsmodels.stats.multitest import multipletests

# Enter exactly 20 p-values here
p_values = [
    0.040923, 0.002909, 0.005913, 0.006961, 0.019257, 0.006221, 0.000013, 0.204400, 0.005049, 0.010564, 0.113417, 0.003611, 0.245672, 0.012980, 0.029142, 0.734651, 0.974925, 0.699055, 0.203012, 0.992319
]


# Benjamini-Hochberg FDR correction
_, fdr_adjusted_p_values, _, _ = multipletests(
    p_values,
    alpha=0.05,
    method="fdr_bh"
)

# Print list
print("FDR-adjusted p-values:")
for p in fdr_adjusted_p_values:
    print(f"{p:.6f}")
```
